```python
#write mode
f = open("Data.txt","w")
content = "I am a text file"
f.write(content)
f.close()
```

```python
f = open("Data.txt","a")
f.writelines(["\nHello world","\nPython lab"])
f.close()
```

```python
f = open("Data.txt","a")
f.write("This is a new Line")
f.close()
```

```python
# read mode
f = open("Data.txt","r")
data = f.read(6)
print(data)
f.tell()
f.seek(2)
print(f.read())
```

```python
with open("Data.txt","r") as f:
   firstline = f.readline()
   print(firstline)
with open("Data.txt","r") as f:
   All_lines = f.readlines()
   print(All_lines)
```
