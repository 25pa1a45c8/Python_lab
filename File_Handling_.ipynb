```python
with open("student.txt", "r") as file:
  data=file.read().split()
  words=[i.upper() for i in data]
  print(words)
  words.sort()
  print(words)
with open("copy.txt","w") as f:
  for i in words:
    f.write(i+"\n")
print("done")
```

```python
#copying content of one file to other
with open("f1.txt","r") as f:
  data=f.read()
  print("reading done")
with open("f2.txt","w") as f2:
  f2.write(data)
  print("file copied")
```

```python
#prgram that prints each line in a file in reverse order
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python
with open("f1.txt","r") as f:
  data=f.read().split()
  print(data)
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("copy.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
print("done")
```
