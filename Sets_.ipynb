```python
# program using mathematical notations on sets (operations)
s1 ={4,1,2,8}
s2 ={3,5,1,4}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
# methods on sets
s = {12,"hello",(3,7,9)}
s.add(23)  #add
print(s)
```

```python
k = {(5,6)}
s.update(k)         #update
print(s)
```

```python
s.remove(12)   #remove
print(s)
```

```python
s.discard("hello")    #discard
print(s)
```

```python
m1 = {2,4,8,6}
m1.pop()  #pop
print(m1)
```

```python
m1.clear()
print(m1)
```

```python
#intersection_update/union_update and union/intersecton
C={5,6,1}
D={9,8,5}
C.intersection_update(D)
print(C)
```

```python
#subsets/supersets/disjoint
C.union(D)
```

```python
C.intersection(D)
```
