```python
#program on arthimatic operatios
a = int(input("Enter a value"))
b = int(input("Enter b value"))
print(a+b)
print(a-b)
print(a*b)
print(a//b)
print(a/b)
print(a%b)
print(a**b)
```

```python
# program on assignment operators
a = 10
a = a+5
a = a-4
a = a*2
a = a/2
print(a)
```

```python
# program on logical operators
a = True
b = False
print(a and b)
print(a or b)
print( not b )
```

```python
# program on comparsion operation
a = True
b = False
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
```

```python
# program on bitwise operators
a = int(input("Enter a value"))
b = int(input("Enter b value"))
print(a & b)
print(a | b)
print(a ^ b)
print( ~ a )
print(a << 1)
print(a >> 1)
```

```python
#program on ternary operator
a = 20
b = 60
largest = a if a>b else b
print("Largest = ",largest)
```

```python
# program on membership operator
list  = [10,14,19,50]
print(48 in list)
print(14 in list)
```

```python
#program on identity operators
a = [4,5]
b = a
c = [4,5]
print(a is b)
print(a is c)
print(a == c)
```
