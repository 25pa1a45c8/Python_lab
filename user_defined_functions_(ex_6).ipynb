```python
# program that checks palindrome or not using functions
def palindrome(num):
  original = num
  rev = 0
  while(num != 0):
    rem = num%10
    rev = rev*10+rem
    num = num//10
  if(rev == original):
    return True
  else:
    return False
print(palindrome(505))
print(palindrome(12345))
```

```python
#program reverse of a string using function
def reverse_str(s):
  rev = " "
  for i in range(len(s)-1,-1,-1):
    rev = rev+s[i]
  return rev
print(reverse_str("john"))
```

```python
# example on variable lenght arguments
def add(*num):
  print(num)
  print(sum(num))
add(1,2,3,4,5,6)
```

```python
# function that returns multiple values
def multiple_values(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},the avg is {res_avg}")
```
