```python
# program to count the number of vowels and consonents of the string
a = input()
b = a.lower()
count_v = 0
count_c = 0
vowels = 'aeiou'
for ch in b:
  if ch.isalpha():
    if ch in vowels:
       count_v +=1
    else:
       count_c +=1
print(count_v)
print(count_c)
```
