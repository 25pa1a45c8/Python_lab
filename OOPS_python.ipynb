```python
class Shape:
  def area():
    pass
  def perimeter():
    pass
class Circle(Shape):
  def __init__(self,r):
    self.r=r
  def area(self):
    return 3.14*self.r*self.r
  def perimeter(self):
    return 2*3.14*self.r
class Square(Shape):
  def __init__(self,side):
    self.side=side
  def area(self):
    return self.side*self.side
  def perimeter(self):
    return 4*self.side
class Triangle(Shape):
  def __init__(self,a,b,c,h):
    self.a=a
    self.b=b
    self.c=c
    self.h=h
  def area(self):
    return 0.5*self.b*self.h
  def perimeter(self):
    return self.a+self.b+self.c
c1=Circle(4)
print("Area of the circle : ",c1.area())
print("Perimeter of the circle : ",c1.perimeter())
s1=Square(5)
print("Area of square : ",s1.area())
print("Perimeter of square : ",s1.perimeter())
t1=Triangle(3,4,5,6)
print("Area of triangle : ",t1.area())
print("Perimeter of triangle : ",t1.perimeter())
```
