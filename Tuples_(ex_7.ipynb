```python
#operations in tuples
t1 = (2,7,8.5,'hello',7)
t2 = (1,3,5,9)
t3 = t1+t2        # concatenation
print(t3)
print(t2*4)       # repetation
for i in t1:      # iteration
  print(i)
print(t1[1:3])    # slicing
print(5 in t2)    # membership
```

```python
#functions in tuples
print(sum(t2))
print(max(t2))
print(min(t2))
print(len(t1))
```

```python
#methods in tuples
print(t1.count(7))    # count
print(t2.index(9))    # indexing
```

```python
#zip
t4 = (2,4,6,8)
print(tuple(zip(t2,t4)))
```

```python
#user defined program using tuple
t = tuple(map(int,input("enter values : ").split()))    # map
print(t)
T = tuple(int(x) for x in input().split())    # Tuple comprehension
print(T)
```
