```python
# finding greatest among three numbers
a = int(input("Enter a value"))
b = int(input("Enter b value"))
c = int(input("Enter c value"))
if(a>b and a>c):
  print(f" {a} is greater")
elif(b>a and b>c):
  print(f" {b} is greater")
else :
  print(f" {c} is greater")
```

```python
# print FIZZ if a number is divisible by 3,print BUZZ if a number is divisible by 5,print FIZZBUZZ if it is divisible by 3 and 5
a = int(input("Enter a Number"))
if(a%3==0 and a%5!=0):
  print("FIZZ")
if(a%5==0 and a%3 !=0):
  print("BUZZ")
if(a%3==0 and a%5==0):
  print("FIZZBUZZ")
else:
  print("Nothing")
```

```python
# program on match
day = int(input("Enter any number"))
match day:
  case 1:
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print ("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid")
```
